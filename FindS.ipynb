```python
import pandas as pd
```

```python
data = {
    "Day": ["D1", "D2", "D3", "D4", "D5"],
    "Outlook": ["Sunny", "Sunny", "Sunny", "Sunny", "Sunny"],
    "Temperature": ["Mild", "Mild", "Mild", "Cool", "Mild"],
    "Humidity": ["Normal", "Normal", "Normal", "High", "High"],
    "Wind": ["Weak", "Weak", "Weak", "Weak", "Weak"],
    "Play": ["Yes", "Yes", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

print(df)
```

```python
features = ['Outlook', 'Temperature', 'Humidity', 'Wind']
target = 'Play'

# Select positive examples
positive_examples_df = df[df[target] == 'Yes']

# Start with NULL hypothesis
h = ['ϕ' for _ in features]
print("Initial Null Hypothesis:")
print(h)

# Find-S algorithm
for i, example in positive_examples_df.iterrows():

    for j, feature in enumerate(features):

        # If hypothesis is NULL, initialize it with first positive example
        if h[j] == 'ϕ':
            h[j] = example[feature]

        # If hypothesis is already generalized, leave it as '?'
        elif h[j] == '?':
            continue

        # If values differ, generalize to '?'
        elif h[j] != example[feature]:
            h[j] = '?'

    print(f"Step {i+1}: Hypothesis {h}")


print("\nMost Specific Hypothesis (Find-S):")
print(h)
```
